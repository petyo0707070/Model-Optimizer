```python
#code 1 ,k fixed?
import numpy as np
import pandas as pd
from joblib import Parallel, delayed

# Reusing previously defined functions
def simulate_arma_panel(N, T, phi, theta, seed):
    np.random.seed(seed)
    alpha = np.random.normal(size=N)
    eps   = np.random.normal(size=(N, T+1))
    y     = np.zeros((N, T+1))
    y[:,0] = alpha/(1-phi)
    for t in range(1, T+1):
        y[:,t] = alpha + phi*y[:,t-1] + eps[:,t] + theta*eps[:,t-1]
    return y[:,1:]

def fe_sieve_estimate(y, p):
    N, T = y.shape
    Y = y[:, p:] - y[:, p:].mean(axis=1, keepdims=True)
    X = np.stack([
        y[:, p-k:-k] - y[:, p:].mean(axis=1, keepdims=True)
        for k in range(1, p+1)
    ], axis=2)
    M = X.reshape(-1, p)
    return np.linalg.pinv(M) @ Y.ravel()

def bcfe_analytic(y, p, phi, theta):
    fe_hat = fe_sieve_estimate(y, p)
    N, T   = y.shape
    Y = y[:, p:] - y[:, p:].mean(axis=1, keepdims=True)
    psi_inf = np.array([
        theta**(k-1)*(theta + (-1)**(k-1)*phi)
        for k in range(1, p+1)
    ])
    B_hat = np.zeros(p)
    for k in range(1, p+1):
        denom = (Y[:, :-k]**2).sum()
        num = 0.0
        for ell in range(1, p+1):
            valid = Y.shape[1] - max(k, ell)
            if valid>0:
                num += psi_inf[ell-1]*(Y[:,ell:ell+valid]*Y[:,k:k+valid]).sum()
        B_hat[k-1] = num/denom if denom>0 else 0.0
    return fe_hat - B_hat/(T-p)

def fixed_design_wild_bootstrap_coefs(y, p, B, phi, theta):
    fe_hat = fe_sieve_estimate(y, p)
    N, T = y.shape
    y_bar = y[:, p:].mean(axis=1, keepdims=True)
    Y = y[:, p:] - y_bar
    X = np.stack([y[:, p-k:-k] - y_bar for k in range(1,p+1)], axis=2)
    resid = Y - (X.reshape(-1,p) @ fe_hat).reshape(N, T-p)
    resid -= resid.mean(axis=1, keepdims=True)
    out = np.zeros((B, p))
    for b in range(B):
        eps_star = resid * np.random.choice([-1,1], size=resid.shape)
        y_star = np.zeros_like(y)
        y_star[:,:p] = y_bar
        for t in range(p, T):
            lag = np.stack([y_star[:,t-k] - y_bar.ravel() for k in range(1,p+1)], axis=1)
            y_star[:,t] = lag.dot(fe_hat) + eps_star[:,t-p]
        out[b] = fe_sieve_estimate(y_star, p)
    return out

def recursive_design_wild_bootstrap_coefs(y, p, B, phi, theta):
    fe_orig = fe_sieve_estimate(y, p)
    N, T = y.shape
    y_bar = y[:, p:].mean(axis=1)
    Y = y[:, p:] - y_bar[:,None]
    X = np.stack([y[:, p-k:-k] - y_bar[:,None] for k in range(1,p+1)], axis=2)
    u = Y - (X.reshape(-1,p) @ fe_orig).reshape(N, T-p)
    nu = np.empty_like(u); nu[:,0] = u[:,0]
    for t in range(1, T-p):
        nu[:,t] = u[:,t] - theta*u[:,t-1]
    def one_boot(_):
        y_star = np.zeros_like(y)
        y_star[:,:p] = y_bar[:,None]
        nu_prev = np.zeros(N)
        for t in range(p, T):
            nu_star = np.array([np.random.choice(nu[i]) for i in range(N)])
            if t>=p:
                hist = y_star[:,:t]
                yb = hist[:, p:].mean(axis=1, keepdims=True)
                Yh = hist[:, p:] - yb
                Xh = np.stack([hist[:, p-k:-k] - yb for k in range(1,p+1)], axis=2)
                fe_t = np.linalg.pinv(Xh.reshape(-1,p)) @ Yh.ravel()
            else:
                fe_t = fe_orig
            lag = np.stack([y_star[:,t-k] for k in range(1,p+1)], axis=1)
            lag_dm = lag - y_bar[:,None]
            y_star[:,t] = lag_dm @ fe_t + nu_star + theta*nu_prev
            nu_prev = nu_star
        return fe_sieve_estimate(y_star, p)
    boots = Parallel(n_jobs=-1)(delayed(one_boot)(b) for b in range(B))
    return np.vstack(boots)

def pair_bootstrap_coefs(y, p, B):
    N, _ = y.shape
    out = np.zeros((B,p))
    for b in range(B):
        idx = np.random.choice(N, N, replace=True)
        out[b] = fe_sieve_estimate(y[idx], p)
    return out

# Monte Carlo across N and T
results = []
phis = (0.6, 0.4)
for N in [25, 50, 100]:
    for T in [25, 50, 100]:
        p = int(np.floor(T**0.25))
        reps, B = 200, 200
        true_sum = sum(phis) / (1 + phis[0])
        sums_acc = {m: [] for m in ['FE','BCFE','FDW','RDW','PB']}
        cov_acc  = {m: [] for m in ['FDW','RDW','PB']}
        for r in range(reps):
            y = simulate_arma_panel(N, T, *phis, seed=r)
            fe = fe_sieve_estimate(y, p); sums_acc['FE'].append(fe.sum())
            bc = bcfe_analytic(y,p,*phis); sums_acc['BCFE'].append(bc.sum())
            fd = 2*fe - fixed_design_wild_bootstrap_coefs(y,p,B,*phis).mean(0)
            sums_acc['FDW'].append(fd.sum())
            rd_boots = recursive_design_wild_bootstrap_coefs(y,p,B,*phis)
            rd = 2*fe - rd_boots.mean(0); sums_acc['RDW'].append(rd.sum())
            pb = 2*fe - pair_bootstrap_coefs(y,p,B).mean(0); sums_acc['PB'].append(pb.sum())
            # coverage
            for key, boots in [('FDW', fixed_design_wild_bootstrap_coefs(y,p,B,*phis)),
                               ('RDW', rd_boots),
                               ('PB', pair_bootstrap_coefs(y,p,B))]:
                bsums = (2*fe - boots).sum(1)
                lo, hi = np.percentile(bsums, [2.5,97.5])
                cov_acc[key].append(int(lo <= true_sum <= hi))
        # aggregate
        row = {'N':N,'T':T}
        for m in sums_acc:
            arr = np.array(sums_acc[m])
            row[f'Bias_{m}'] = arr.mean() - true_sum
            row[f'RMSE_{m}'] = np.sqrt(((arr-true_sum)**2).mean())
        for m in cov_acc:
            row[f'Coverage_{m}'] = np.mean(cov_acc[m])
        results.append(row)

df = pd.DataFrame(results)
# save to Excel
df.to_excel('/mnt/data/simulation_results.xlsx', index=False)
```

```python
#code 2 tryiing to extend to growing k,ar(k) model
import numpy as np
import pandas as pd
from joblib import Parallel, delayed

# --------------------------------------------------------------------
# 1) New DGP: AR(K_T) with K_T = floor(T**0.25)
# --------------------------------------------------------------------
def simulate_ar_panel(N, T, alphas, seed):
    """
    Simulate an N×T panel from
       y_{i,t} = eta_i + sum_{k=1}^K alphas[k-1] * y_{i,t-k} + eps_{i,t}
    where eps ~ N(0,1), eta_i ~ N(0,1), K = len(alphas).
    """
    np.random.seed(seed)
    K = len(alphas)
    eta = np.random.normal(size=N)
    eps = np.random.normal(size=(N, T+K))
    y   = np.zeros((N, T+K))
    # initialize at unconditional mean = eta / (1 - sum(alphas))
    mu = eta / (1 - alphas.sum())
    y[:, :K] = mu[:, None]
    for t in range(K, T+K):
        lagged = np.stack([y[:, t-k] for k in range(1, K+1)], axis=1)  # (N, K)
        y[:, t] = eta + (lagged * alphas).sum(axis=1) + eps[:, t]
    return y[:, K:]

# --------------------------------------------------------------------
# 2) FE‐sieve + Bootstrap routines (unchanged)
# --------------------------------------------------------------------
def fe_sieve_estimate(y, p):
    N, T = y.shape
    Y = y[:, p:] - y[:, p:].mean(axis=1, keepdims=True)
    X = np.stack([
        y[:, p-k:-k] - y[:, p:].mean(axis=1, keepdims=True)
        for k in range(1, p+1)
    ], axis=2)
    M = X.reshape(-1, p)
    return np.linalg.pinv(M) @ Y.ravel()

def fixed_design_wild_bootstrap_coefs(y, p, B):
    fe_hat = fe_sieve_estimate(y, p)
    N, T = y.shape
    y_bar = y[:, p:].mean(axis=1, keepdims=True)
    Y = y[:, p:] - y_bar
    X = np.stack([y[:, p-k:-k] - y_bar for k in range(1,p+1)], axis=2)
    resid = Y - (X.reshape(-1,p) @ fe_hat).reshape(N, T-p)
    resid -= resid.mean(axis=1, keepdims=True)
    out = np.zeros((B, p))
    for b in range(B):
        eps_star = resid * np.random.choice([-1,1], size=resid.shape)
        y_star = np.zeros_like(y)
        y_star[:,:p] = y_bar
        for t in range(p, T):
            lag = np.stack([y_star[:,t-k] - y_bar.ravel() for k in range(1,p+1)], axis=1)
            y_star[:,t] = lag.dot(fe_hat) + eps_star[:,t-p]
        out[b] = fe_sieve_estimate(y_star, p)
    return out

def recursive_design_wild_bootstrap_coefs(y, p, B):
    fe_orig = fe_sieve_estimate(y, p)
    N, T = y.shape
    y_bar = y[:, p:].mean(axis=1)
    Y = y[:, p:] - y_bar[:,None]
    X = np.stack([y[:, p-k:-k] - y_bar[:,None] for k in range(1,p+1)], axis=2)
    u = Y - (X.reshape(-1,p) @ fe_orig).reshape(N, T-p)
    nu = np.empty_like(u); nu[:,0] = u[:,0]
    for t in range(1, T-p):
        # here θ=0 (pure AR), so nu = u
        nu[:,t] = u[:,t]
    def one_boot(_):
        y_star = np.zeros_like(y)
        y_star[:,:p] = y_bar[:,None]
        nu_prev = np.zeros(N)
        for t in range(p, T):
            nu_star = np.array([np.random.choice(nu[i]) for i in range(N)])
            if t>=p:
                hist = y_star[:,:t]
                yb = hist[:, p:].mean(axis=1, keepdims=True)
                Yh = hist[:, p:] - yb
                Xh = np.stack([hist[:, p-k:-k] - yb for k in range(1,p+1)], axis=2)
                fe_t = np.linalg.pinv(Xh.reshape(-1,p)) @ Yh.ravel()
            else:
                fe_t = fe_orig
            lag = np.stack([y_star[:,t-k] for k in range(1,p+1)], axis=1)
            lag_dm = lag - y_bar[:,None]
            y_star[:,t] = lag_dm @ fe_t + nu_star
        return fe_sieve_estimate(y_star, p)
    boots = Parallel(n_jobs=-1)(delayed(one_boot)(b) for b in range(B))
    return np.vstack(boots)

def pair_bootstrap_coefs(y, p, B):
    N, _ = y.shape
    out = np.zeros((B,p))
    for b in range(B):
        idx = np.random.choice(N, N, replace=True)
        out[b] = fe_sieve_estimate(y[idx], p)
    return out

# --------------------------------------------------------------------
# 3) Monte Carlo with growing K_T = p = floor(T**0.25)
# --------------------------------------------------------------------
results = []
reps, B = 200, 200

for N in [25, 50, 100]:
    for T in [25, 50, 100]:
        p = int(np.floor(T**0.25))   # sieve order
        K = p                        # true AR order grows with T
        alphas = np.full(K, 0.2)     # true AR coefficients
        true_sum = alphas.sum()

        sums_acc = {m: [] for m in ['FE','FDW','RDW','PB']}
        cov_acc  = {m: [] for m in ['FDW','RDW','PB']}

        for r in range(reps):
            # simulate AR(K) panel
            y = simulate_ar_panel(N, T, alphas, seed=r)

            # FE‐sieve
            fe = fe_sieve_estimate(y, p)
            sums_acc['FE'].append(fe.sum())

            # bootstrap‐bias corrections
            fdw = 2*fe - fixed_design_wild_bootstrap_coefs(y, p, B).mean(0)
            rdw = 2*fe - recursive_design_wild_bootstrap_coefs(y, p, B).mean(0)
            pb  = 2*fe - pair_bootstrap_coefs(y, p, B).mean(0)

            sums_acc['FDW'].append(fdw.sum())
            sums_acc['RDW'].append(rdw.sum())
            sums_acc['PB'].append(pb.sum())

            # coverage
            for key, boots in [('FDW', fixed_design_wild_bootstrap_coefs(y,p,B)),
                               ('RDW', recursive_design_wild_bootstrap_coefs(y,p,B)),
                               ('PB', pair_bootstrap_coefs(y,p,B))]:
                bsums = (2*fe - boots).sum(1)
                lo, hi = np.percentile(bsums, [2.5,97.5])
                cov_acc[key].append(int(lo <= true_sum <= hi))

        # aggregate
        row = {'N':N, 'T':T, 'K':K}
        for m in sums_acc:
            arr = np.array(sums_acc[m])
            row[f'Bias_{m}'] = arr.mean() - true_sum
            row[f'RMSE_{m}'] = np.sqrt(((arr-true_sum)**2).mean())
        for m in cov_acc:
            row[f'Coverage_{m}'] = np.mean(cov_acc[m])

        results.append(row)

df = pd.DataFrame(results)
print(df)
df.to_excel('/mnt/data/simulation_results_growingK.xlsx', index=False)
```
