```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# --- 1. Simulation & estimators (same as before) ---

def simulate_arma_panel(N, T, phi, theta_ma, seed=None):
    if seed is not None:
        np.random.seed(seed)
    alpha = np.random.normal(size=N)
    eps   = np.random.normal(size=(N, T+1))
    y     = np.zeros((N, T+1))
    y[:, 0] = alpha / (1 - phi)
    for t in range(1, T+1):
        y[:, t] = alpha + phi*y[:, t-1] + eps[:, t] + theta_ma*eps[:, t-1]
    return y[:, 1:]

def fe_sieve_estimate(y, K):
    N, T = y.shape
    Y = y[:, K:]
    X = np.stack([y[:, K-k:-k] for k in range(1, K+1)], axis=2)
    Y_bar = Y.mean(axis=1, keepdims=True)
    X_bar = X.mean(axis=1, keepdims=True)
    Y_w = (Y - Y_bar).reshape(-1)
    X_w = (X - X_bar).reshape(-1, K)
    return sm.OLS(Y_w, X_w).fit().params

# (Re‐use your corrected bcfe_estimate, rd_bootstrap_estimate, fd_bootstrap_estimate, pb_estimate, compute_bias_rmse)
def bcfe_estimate(y, K, true_betas):
    # analytic first-order bias correction (Lee et al.)
    N, T = y.shape; p = K
    Y = y[:, p:]; Y_bar = Y.mean(axis=1, keepdims=True)
    Y_tilde = Y - Y_bar
    theta_fe = fe_sieve_estimate(y, p)
    b_lee = np.zeros(p)
    J = T - p
    for k in range(1, p+1):
        num = 0.0; denom = 0.0
        for i in range(N):
            for j in range(J):
                ik = j - k
                if ik>=0:
                    denom += Y_tilde[i, ik]**2
        for ell in range(1, p+1):
            beta_ell = true_betas[ell-1]
            for i in range(N):
                for j in range(J):
                    ik = j - k; iℓ = j - ell
                    if ik>=0 and iℓ>=0:
                        num += beta_ell * Y_tilde[i, iℓ] * Y_tilde[i, ik]
        b_lee[k-1] = num / denom / T
    return theta_fe - b_lee

def rd_bootstrap_estimate(y, K, B=500):
    # recursive-design residual bootstrap
    N, T = y.shape; p = K
    ybar = y[:, p:].mean(axis=1)
    theta_fe = fe_sieve_estimate(y, p)
    Y = y[:, p:]; X = np.stack([y[:, p-k:-k] for k in range(1, p+1)], axis=2)
    Y_bar = Y.mean(axis=1, keepdims=True)
    X_bar = X.mean(axis=1, keepdims=True)
    resid = (Y - Y_bar) - np.tensordot(X - X_bar, theta_fe, axes=([2],[0]))
    coefs = []
    for _ in range(B):
        nu = np.random.choice([-1,1], size=resid.shape)
        u_star = resid * nu
        y_star = np.zeros_like(y); y_star[:, :p] = y[:, :p]
        for t in range(p, T):
            lagged = np.stack([y_star[:, t-k] for k in range(1, p+1)], axis=1)
            y_star[:, t] = ybar + (lagged - ybar[:, None]) @ theta_fe + u_star[:, t-p]
        coefs.append(fe_sieve_estimate(y_star, p))
    return np.mean(coefs, axis=0)

def fd_bootstrap_estimate(y, K, B=500):
    # fixed-design residual bootstrap
    N, T = y.shape; p = K
    ybar = y[:, p:].mean(axis=1)
    theta_fe = fe_sieve_estimate(y, p)
    X_orig = np.stack([y[:, p-k:-k] for k in range(1, p+1)], axis=2)
    Y = y[:, p:]; Y_bar = Y.mean(axis=1, keepdims=True)
    X_bar = X_orig.mean(axis=1, keepdims=True)
    resid = (Y - Y_bar) - np.tensordot(X_orig - X_bar, theta_fe, axes=([2],[0]))
    coefs = []
    for _ in range(B):
        nu = np.random.choice([-1,1], size=resid.shape)
        u_star = resid * nu
        y_star = np.zeros_like(y); y_star[:, :p] = y[:, :p]
        for t in range(p, T):
            lagged_orig = X_orig[:, t-p, :]
            centered = lagged_orig - ybar[:, None]
            y_star[:, t] = ybar + centered @ theta_fe + u_star[:, t-p]
        coefs.append(fe_sieve_estimate(y_star, p))
    return np.mean(coefs, axis=0)

def pb_estimate(y, K, B=500):
    # pairs bootstrap
    N, _ = y.shape; p = K
    coefs = []
    for _ in range(B):
        idx = np.random.choice(N, N, replace=True)
        coefs.append(fe_sieve_estimate(y[idx, :], p))
    return np.mean(coefs, axis=0)

def compute_bias_rmse(est_list, true_vals):
    arr = np.array(est_list); true = np.array(true_vals)
    bias = arr.mean(axis=0) - true
    rmse = np.sqrt(((arr - true)**2).mean(axis=0))
    return bias, rmse

# --- 2. Loop over (N,T) grid, fixed phi/theta, collect all lags in long format ---

Ns = [50, 100, 200]
Ts = [25, 50, 100]
phi, theta_ma = 0.6, 0.4

n_rep, B = 1000, 500   # start small for speed, then increase

rows = []
for N in Ns:
    for T in Ts:
        K = int(np.floor(T**0.25))
        # True infinite‐AR coefficients β₁…βₖ
        true_betas = [phi + theta_ma] + [phi**(k-1)*theta_ma for k in range(2, K+1)]

        # Storage per method
        results = {m: [] for m in ['FE','BCFE','RD','FD','PB']}
        for rep in range(n_rep):
            y = simulate_arma_panel(N, T, phi, theta_ma, seed=rep)
            results['FE'].append(fe_sieve_estimate(y, K))
            results['BCFE'].append(bcfe_estimate(y, K, true_betas))
            results['RD'].append(rd_bootstrap_estimate(y, K, B))
            results['FD'].append(fd_bootstrap_estimate(y, K, B))
            results['PB'].append(pb_estimate(y, K, B))

        # Compute bias & RMSE for each lag and method
        for method, est_list in results.items():
            bias, rmse = compute_bias_rmse(est_list, true_betas)
            for lag in range(1, K+1):
                rows.append({
                    'N': N,
                    'T': T,
                    'Method': method,
                    'Lag': lag,
                    'Bias': bias[lag-1],
                    'RMSE': rmse[lag-1]
                })

# --- 3. Export to Excel ---

df = pd.DataFrame(rows)
df.to_excel('bias_rmse_all_lags.xlsx', index=False)
print("Saved bias/RMSE for all AR lags to bias_rmse_all_lags.xlsx")
```
